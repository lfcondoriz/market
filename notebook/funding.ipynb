## 2) category, symbol, start_time

si mandás startTime, también necesitás endTime

```python
start_time = "2025-01-15 00:00:00"
end_time = "2025-01-31 23:59:59"


start_time = pd.to_datetime(start_time, utc=True)
start_time_ms = int(start_time.timestamp() * 1000)

end_time = pd.to_datetime(end_time, utc=True)
end_time_ms = int(end_time.timestamp() * 1000)
```

```python
from pybit.unified_trading import HTTP
session = HTTP()
data = session.get_funding_rate_history(
    category="linear",
    symbol="BTCUSDT",
    startTime=start_time_ms,
    endTime=end_time_ms
)
```

```python
df = pd.DataFrame(data["result"]["list"])

df["fundingRateTimestamp"] = pd.to_datetime(
    pd.to_numeric(df["fundingRateTimestamp"]),
    unit="ms",
    utc=True
)
```

```python
df
```
